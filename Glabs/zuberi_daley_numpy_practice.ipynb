This notebook covers NumPy data type objects, one-dimensional arrays, and common NumPy operations (element-wise math, dot product, and summary statistics).

```python
# Import NumPy using its standard alias
import numpy as np

print('NumPy version:', np.__version__)
```

## Task 1: Data Type Objects in NumPy
The `dtype` argument tells NumPy exactly how to store each element in memory. All elements in a NumPy array share one data type.

```python
# 1. Array stored as 32-bit integers
int_arr = np.array([10, 20, 30, 40, 50], dtype='int32')

print('Array:', int_arr)
print('Data type:', int_arr.dtype)
```

```python
# 2. Array stored as 64-bit floating-point numbers
float_arr = np.array([1.5, 2.75, 3.0, 4.25, 5.5], dtype='float64')

print('Array:', float_arr)
print('Data type:', float_arr.dtype)
```

## Task 2: One-Dimensional Array

```python
# Create a 1D array of the integers 1 through 5
# np.arange(start, stop) excludes the stop value, so we use 6
one_d = np.arange(1, 6)

print('1D array:', one_d)
print('Shape:', one_d.shape)   # (5,) -> one dimension with 5 elements
print('Dimensions:', one_d.ndim)
```

## Task 3: NumPy Operations

```python
# 1. Create two arrays of the same shape for element-wise operations
arr1 = np.array([4, 8, 15, 16, 23])
arr2 = np.array([2, 4, 5, 8, 10])

print('arr1:', arr1)
print('arr2:', arr2)

# Element-wise operations apply to matching positions in each array
print('\nAddition:      ', arr1 + arr2)
print('Subtraction:   ', arr1 - arr2)
print('Multiplication:', arr1 * arr2)
print('Division:      ', arr1 / arr2)
```

```python
# 2. Dot product: multiply matching elements, then sum the results
# (4*2) + (8*4) + (15*5) + (16*8) + (23*10)
dot_product = np.dot(arr1, arr2)

print('Dot product of arr1 and arr2:', dot_product)
```

```python
# 3. Summary statistics for arr1
mean_val = np.mean(arr1)      # average of all values
median_val = np.median(arr1)  # middle value when sorted
std_val = np.std(arr1)        # population standard deviation (ddof=0 by default)

print(f'Mean:               {mean_val:.2f}')
print(f'Median:             {median_val:.2f}')
print(f'Standard deviation: {std_val:.2f}')
```
